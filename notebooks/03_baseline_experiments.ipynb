Benchmarking Random Forest, CNN (images only), and Bi-LSTM (light curves only) against the Cross-Attention Multimodal Model.

```python
import sys; sys.path.append('..')
from src.data.dataset_builder import generate_benchmark_events
from src.data.splitter import ObjectLevelSplitter
from src.baselines.evaluate_baselines import BaselineEvaluator

events = generate_benchmark_events(num_known=120, num_anomalies=20)
known_classes = ['SN_Ia', 'SN_II', 'Stellar_Flare', 'Variable_Star']
splitter = ObjectLevelSplitter(known_classes, ['LRN', 'SLSN', 'TDE'])
splits = splitter.split_events(events)

train_events = [e for e in events if e.object_id in splits.train_ids]
test_events = [e for e in events if e.object_id in splits.test_known_ids]

evaluator = BaselineEvaluator(known_classes)
results = evaluator.run_benchmark(train_events, test_events)
```
