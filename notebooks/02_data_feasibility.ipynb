# 02. Data Feasibility & Object-Level Splitting (Stage 0 & 1)
Ensuring zero observation leakage by performing rigorous object-level splitting into known classes and held-out rare classes.

```python
import sys; sys.path.append('..')
from src.data.dataset_builder import generate_benchmark_events
from src.data.splitter import ObjectLevelSplitter

events = generate_benchmark_events(num_known=100, num_anomalies=20)
splitter = ObjectLevelSplitter(
    known_classes=['SN_Ia', 'SN_II', 'Stellar_Flare', 'Variable_Star'],
    held_out_classes=['LRN', 'SLSN', 'TDE']
)
splits = splitter.split_events(events)
print('Train objects:', len(splits.train_ids))
print('Val objects:', len(splits.val_ids))
print('Test Known objects:', len(splits.test_known_ids))
print('Held-out Anomaly objects:', len(splits.test_anomaly_ids))
```
